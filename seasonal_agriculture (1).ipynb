```python
#CHART 2 STATE VS CROP WHICH STATE GROWING WHICH CROP
import pandas as pd
import matplotlib.pyplot as plt

# Create cross-tabulation
crop_state = pd.crosstab(df['State'], df['Crop'])

# Create stacked bar chart
crop_state.plot(kind='bar', stacked=True, figsize=(12, 6), colormap='tab20')

plt.title('State vs Crop Distribution', fontsize=16, fontweight='bold')
plt.xlabel('State', fontsize=12)
plt.ylabel('Number of Crops', fontsize=12)
plt.legend(title='Crop', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# TOP 3 STATES FOR EACH CROP
import pandas as pd

# Create cross-tabulation
crop_state = pd.crosstab(df['State'], df['Crop'])

print("="*60)
print("TOP 3 STATES FOR EACH CROP")
print("="*60)

for crop in crop_state.columns:
    top_states = crop_state[crop].nlargest(3)
    print(f"\n{crop}:")
    for state, count in top_states.items():
        print(f"  {state}: {count}")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Calculate correlation between nutrients and yield
correlation = df[['Nitrogen_kg_ha', 'Phosphorus_kg_ha', 'Potassium_kg_ha', 'Yield_Tonnes_Ha']].corr()

print("="*60)
print("CORRELATION BETWEEN NUTRIENTS AND YIELD")
print("="*60)
print(correlation['Yield_Tonnes_Ha'].sort_values(ascending=False))

# Visualize correlation
plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, fmt='.2f', cmap='coolwarm', linewidths=1)
plt.title('Correlation Between Nutrients and Yield', fontsize=16, fontweight='bold')
plt.show()
```

```python
#FACTORS AFFECTING CROP PRODUCTION
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Select factors
factors = ['Rainfall_mm', 'Avg_Temperature_C', 'Humidity_pct',
           'Sunlight_Hours_Day', 'Soil_pH', 'Soil_Moisture_pct',
           'Yield_Tonnes_Ha']

# Calculate correlation
correlation_matrix = df[factors].corr()

# Create heatmap
plt.figure(figsize=(12, 10))
mask = np.triu(np.ones_like(correlation_matrix, dtype=bool))
sns.heatmap(correlation_matrix,
            annot=True,
            fmt='.2f',
            cmap='RdBu_r',
            vmin=-1,
            vmax=1,
            center=0,
            mask=mask,
            linewidths=0.5,
            square=True,
            cbar_kws={'label': 'Correlation Coefficient', 'shrink': 0.8})

plt.title('Correlation Heatmap: Environmental Factors vs Yield',
          fontsize=18, fontweight='bold', pad=20)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create box plot with individual points
plt.figure(figsize=(10, 6))

# Box plot
sns.boxplot(data=df, x='Season', y='Yield_Tonnes_Ha',
            palette=['#FF6B6B', '#4ECDC4', '#45B7D1'],
            width=0.6)

# Add swarm plot for individual data points
sns.swarmplot(data=df, x='Season', y='Yield_Tonnes_Ha',
              color='black', alpha=0.5, size=4)

# Add mean values
means = df.groupby('Season')['Yield_Tonnes_Ha'].mean()
for i, season in enumerate(df['Season'].unique()):
    plt.text(i, means[season] + 0.2, f'Mean: {means[season]:.2f}',
             ha='center', va='bottom', fontsize=10, fontweight='bold',
             bbox=dict(boxstyle="round,pad=0.3", facecolor='yellow', alpha=0.7))

plt.title('Yield Distribution by Season', fontsize=16, fontweight='bold')
plt.xlabel('Season', fontsize=12)
plt.ylabel('Yield (Tonnes/Ha)', fontsize=12)
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Which season is the most profitable?
df['Profit_per_Ha'] = df['Profit_INR'] / df['Farm_Area_Hectares']
season_summary = df.groupby('Season').agg({
    'Profit_INR': 'sum',
    'Yield_Tonnes_Ha': 'mean',
    'Market_Price_INR_Tonne': 'mean',
    'Profit_per_Ha': 'mean'
}).round(2)

# Plot
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1']

# Total Profit
season_summary['Profit_INR'].plot(kind='bar', ax=axes[0], color=colors)
axes[0].set_title('Total Profit by Season', fontsize=14, fontweight='bold')
axes[0].set_ylabel('Profit (INR)')
axes[0].grid(axis='y', alpha=0.3)

# Profit per Hectare
season_summary['Profit_per_Ha'].plot(kind='bar', ax=axes[1], color=colors)
axes[1].set_title('Profit per Hectare by Season', fontsize=14, fontweight='bold')
axes[1].set_ylabel('Profit/Ha (INR)')
axes[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
print(season_summary)
```

```python
#How do environmental factors affect yield?
factors = ['Rainfall_mm', 'Avg_Temperature_C', 'Humidity_pct',
           'Sunlight_Hours_Day', 'Yield_Tonnes_Ha']

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for i, season in enumerate(df['Season'].unique()):
    corr = df[df['Season']==season][factors].corr()
    sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdBu_r',
                vmin=-1, vmax=1, center=0, ax=axes[i])
    axes[i].set_title(f'{season} Season', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()

# Show yield correlations only
corr_with_yield = pd.DataFrame()
for season in df['Season'].unique():
    corr = df[df['Season']==season][factors].corr()['Yield_Tonnes_Ha'].drop('Yield_Tonnes_Ha')
    corr_with_yield[season] = corr
print(corr_with_yield.round(3))
```

```python
# Resource comparison for each crop
resources = ['Water_Used_m3', 'Fertilizer_kg_ha', 'Pesticide_Litre_ha', 'Water_Efficiency_t_per_1000m3']
resource_summary = df.groupby('Season')[resources].mean().round(2)

# Plot all resources
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1']

for i, col in enumerate(resources):
    ax = axes[i//2, i%2]
    resource_summary[col].plot(kind='bar', ax=ax, color=colors)
    ax.set_title(col.replace('_', ' '), fontsize=12, fontweight='bold')
    ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
print(resource_summary)
```

```python
# Best crop-irrigation combinations?
fig, axes = plt.subplots(1, 3, figsize=(15, 6))

for i, season in enumerate(df['Season'].unique()):
    pivot = df[df['Season']==season].pivot_table(
        index='Crop',
        columns='Irrigation_Method',
        values='Profit_INR',
        aggfunc='sum'
    ).fillna(0) / 1e6  # Convert to millions

    sns.heatmap(pivot, annot=True, fmt='.1f', cmap='YlOrRd',
                linewidths=0.5, ax=axes[i],
                cbar_kws={'label': 'Profit (INR Millions)'})
    axes[i].set_title(f'{season} Season', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()

# Best combination per season
for season in df['Season'].unique():
    data = df[df['Season']==season]
    best = data.groupby(['Crop', 'Irrigation_Method'])['Profit_INR'].sum().idxmax()
    profit = data.groupby(['Crop', 'Irrigation_Method'])['Profit_INR'].sum().max()
    print(f"{season}: {best[0]} + {best[1]} (₹{profit:,.0f})")
```
